```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
# Load dataset
df = pd.read_csv(r"C:\Users\Vignesh\Downloads\Mall_Customers.csv")

# Display first 5 rows
print(df.head())
```

```python
# Select features
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

# Elbow Method
inertia = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)
    inertia.append(model.inertia_)
```

```python
plt.plot(range(1, 11), inertia, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```
